```python
from selectolax.parser import HTMLParser
from selectolax.lexbor import LexborHTMLParser
import requests
import pandas as pd
```

```python
url = "https://quotes.toscrape.com/"

response = requests.get(url)

soup = HTMLParser(response.text)

quotes = soup.css("div.quote")

for quote in quotes:
    title = quote.css_first("span.text").text(strip=True)
    author = quote.css_first("small.author").text(strip=True)
    tags = [t.text(strip=True) for t in quote.css("a.tag")]

    print(title, author, tags)
```

```python
url = "https://books.toscrape.com/"

response = requests.get(url)

lax = HTMLParser(response.text)

books = lax.css("article.product_pod")

for book in books:
    title = book.css_first("h3 a").attributes["title"]
    rating = book.css_first("p").attributes["class"].split()[1]
    img = url + book.css_first("img").attrs["src"]
    print(img)
```
